```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv("student_data.csv")
```

```python
df.head()
```

```python
df.info()
```

```python
df.describe()
```

```python
df.drop("Unnamed: 0", axis = 1)
```

```python
df.isnull().sum()
```

```python
(df.isnull().mean() * 100).sort_values(ascending=False)
```

```python
df["ParentEduc"].fillna(df["ParentEduc"].mode()[0], inplace=True)
```

```python
df["TestPrep"].fillna(df["TestPrep"].mode()[0], inplace=True)
```

```python
df["ParentMaritalStatus"].fillna(
    df["ParentMaritalStatus"].mode()[0],
    inplace=True
)
```

```python
df["PracticeSport"].fillna(
    df["PracticeSport"].mode()[0],
    inplace=True
)
```

```python
df["IsFirstChild"].fillna(
    df["IsFirstChild"].mode()[0],
    inplace=True
)
```

```python
df["TransportMeans"].fillna(
    df["TransportMeans"].mode()[0],
    inplace=True
)
```

```python
df["WklyStudyHours"].fillna(
    df["WklyStudyHours"].mode()[0],
    inplace=True
)
```

```python
df.head(10)
```

```python
df.isnull().sum()
```

```python
df["NrSiblings"].fillna(
    df["NrSiblings"].mode()[0],
    inplace=True
)
```

```python
df.isnull().sum()
```

```python
df["WklyStudyHours"]=df["WklyStudyHours"].str.replace("05-Oct", "05-10")
```

```python
df.head()
```

```python
# gender distribution

plt.figure(figsize=(8, 5))

ax = sns.countplot(
    data=df,
    x="Gender",
    hue="Gender",
    palette="Set2",
    legend=False
)

for container in ax.containers:
    ax.bar_label(container, fontsize=11, fontweight="bold")

plt.title("Gender Distribution", fontsize=16, fontweight="bold")
plt.xlabel("Gender", fontsize=12)
plt.ylabel("Number of People", fontsize=12)

plt.grid(axis="y", linestyle="--", alpha=0.3)
plt.tight_layout()

plt.show()
```

```python
s = df.groupby("ParentEduc").agg({
    "MathScore": "mean",
    "ReadingScore": "mean",
    "WritingScore":"mean"
})
```

```python
s
```

```python
s.plot(
    kind="bar",
    figsize=(10, 6)
)

plt.title("Average Student Scores by Parent Education")
plt.xlabel("Parent Education")
plt.ylabel("Average Score")
plt.xticks(rotation=45)
plt.legend(title="Subject")
plt.grid(axis="y", linestyle="--", alpha=0.3)
plt.tight_layout()

plt.show()
```

```python
s1 = df.groupby("ParentMaritalStatus").agg({
    "MathScore": "mean",
    "ReadingScore": "mean",
    "WritingScore":"mean"
})
```

```python
s1
```

```python
plt.figure(figsize=(8, 5))

sns.heatmap(
    s1,
    annot=True,
    fmt=".1f",
    cmap="YlGnBu"
)

plt.title("Average Scores by Parent Marital Status")
plt.xlabel("Subject")
plt.ylabel("Parent Marital Status")

plt.tight_layout()
plt.show()
```

```python

fig, ax = plt.subplots(1, 3, figsize=(12, 4))

ax[0].boxplot(df["MathScore"])
ax[0].set_title("Math Score")

ax[1].boxplot(df["ReadingScore"])
ax[1].set_title("Reading Score")

ax[2].boxplot(df["WritingScore"])
ax[2].set_title("Writing Score")

plt.tight_layout()
plt.show()
```
